## 🚻 Gender Distribution

```python
plt.figure(figsize=(5, 5))

sns.countplot(
    data=df,
    x="gender",
    hue="gender",
    palette="Set2",
    legend=False,
    width=0.4,
)


plt.title("Distribution of Students by Gender", fontsize=12, fontweight="bold", pad=15)
plt.xlabel("Gender", fontsize=10, labelpad=10)
plt.ylabel("Number of Students", fontsize=10, labelpad=10) 

plt.tight_layout()
plt.savefig("../images/gender_distribution.png", dpi = 300, bbox_inches = "tight")
plt.show()
```

### 🔍 Key Insights & Observations (Gender Distribution)

* **Balanced Dataset:** The dataset exhibits a highly balanced distribution between male and female students, ensuring that statistical insights are not skewed toward a specific gender category.

* **Higher Female Representation:** Female students represent the majority slice of the dataset with **51.8% (518 students)**, slightly outnumbering male students who stand at **48.2% (482 students)**.

* **Data Validity:** With a total sample size of exactly **1,000 observations**, this distribution provides a robust foundation for conducting subsequent cross-feature analysis (such as analyzing scores based on parental education or lunch types).


## 🍱 Lunch Type Distribution

```python
plt.figure(figsize=(5,5))

sns.countplot(data = df, 
              x = "lunch",
              hue = "lunch",
              palette = "Set2",
              legend = False,
              width = 0.4,
)
plt.title("Lunch Distribution among students", fontsize = 12, fontweight = "bold", pad = 15)

plt.xlabel("Lunch Type", fontsize = 10, labelpad = 10)
plt.ylabel("Number of Students", fontsize = 10, labelpad = 10)

plt.tight_layout()
plt.savefig("../images/lunch_distribution.png", dpi = 300, bbox_inches = "tight")
plt.show()
```

### 🔍 Key Insights & Observations (Lunch Type Distribution)

* **Standard Plan Dominance:** The majority of students in the dataset are enrolled in the **Standard Lunch Plan**, accounting for **64.5% (645 students)** of the total sample size.
* **Subsidy Representation:** Students receiving **Free/Reduced Lunch** make up **35.5% (355 students)** of the population, providing a substantial group size to analyze the socio-economic impact on academic performance.
* **Socio-Economic Indicator:** This feature serves as a critical proxy for socioeconomic status, allowing us to correlate nutrition and financial assistance plans with student scores across math, reading, and writing columns.


## 👨‍👩‍👧 Parental Level of Education

```python
plt.figure(figsize=(10,5))

sns.countplot(data = df, 
              y = "parental level of education",
              hue = "parental level of education",
              palette = "Set2",
              legend = False,
              width = 0.9,
)

plt.title("Parental Level of Education", fontsize = 12, fontweight = "bold", pad = 15)

plt.ylabel("Education Qualification", fontsize = 10, labelpad = 10)
plt.xlabel("Number of Parents", fontsize = 10, labelpad = 10)

plt.tight_layout()
plt.savefig("../images/parental_level_of_education.png", dpi = 300, bbox_inches = "tight")
plt.show()
```

### 🔍 Key Insights & Observations (Parental Education Distribution)

* **Highest Representation:** The majority of parents have completed **'some college'** or hold an **'associate's degree'**, making up the largest educational background segments in the dataset.
* **Lowest Representation:** Parents with a **'master's degree'** represent the absolute minority in this population slice, showing that advanced postgraduate education is rare among the student cohort.
* **Stratification Layer:** This distinct variation in parental background provides an excellent metric to analyze whether higher parental degrees directly correlate with top-tier student overall scores.


## 🏫 Test Preparation Course Distribution

```python
plt.figure(figsize=(5,5))

sns.countplot(data = df, 
              x = "test preparation course",
              hue = "test preparation course",
              palette = "Set2",
              legend = False,
              width = 0.4,
)
plt.title("Test Preparation Course Distribution", fontsize = 12, fontweight = "bold", pad = 15)

plt.xlabel("Preparation Course Status", fontsize = 10, labelpad = 10)
plt.ylabel("Number of Students", fontsize = 10, labelpad = 10)

plt.tight_layout()
plt.savefig("../images/test_preparation.png", dpi = 300, bbox_inches = "tight")
plt.show()
```

### 🔍 Key Insights & Observations (Test Preparation Course)

* **Unprepared Majority:** The vast majority of students in the dataset did **not** take any test preparation course, with **none** being the dominant category.
* **Completed Minority:** Only a smaller subset of the student population successfully completed the preparation course before the examinations.

This variable will later be compared with student scores to determine whether completing a preparation course is associated with better academic performance.
