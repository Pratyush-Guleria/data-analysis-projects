##  📊 Correlation Heatmap for Student Academic Scores

```python
plt.figure(figsize=(10, 5))

scores_columns = ["math score", "reading score", "writing score"]
correlation_matrix = df[scores_columns].corr()


sns.heatmap(
    correlation_matrix, 
    annot=True,          # Displays the exact correlation numbers inside boxes
    cmap="RdYlBu",       # Distinct color gradient (Red/Yellow/Blue) to show strong vs weak relationships
    fmt=".3f",           # Shows numbers precisely up to 3 decimal places
    vmin=-1, vmax=1,     # Keeps the standard mathematical correlation scale bounds
    linewidths=1,        # Clear white spacing between boxes for crisp visibility
    square=True         # Forces boxes to maintain a perfect square aspect ratio
)


plt.title(
    "Statistical Correlation Analysis: Math, Reading, & Writing Test Scores", 
    fontsize=11, 
    fontweight="bold", 
    pad=20
)

plt.tight_layout()

plt.savefig("../images/correlation_heatmap_student_academic_scores.png", dpi = 300, bbox_inches = "tight")
plt.show()
```

### 📊 Analytical Inference:
* **Strong Literacy Synergy:** There is an exceptionally high positive correlation (~0.95) between **Reading** and **Writing** scores, showing that language proficiencies are heavily interdependent.
* **Balanced Math Relationship:** **Math scores** display a strong positive correlation (~0.80) with both literacy metrics, meaning academic performance remains uniform across analytical and linguistic subjects.
* **Conclusion:** Yes, a significant direct linear relationship exists across all three disciplines. A high score in one domain serves as a highly reliable statistical predictor of success in the other two.

## 📊 Do Math Toppers Excel in Reading?

```python
plt.figure(figsize = (6, 5))

sns.regplot(
    data = df,
    x = "math score",
    y = "reading score",
    scatter_kws={"color": "royalblue", "alpha": 0.6, "s": 50}, 
    line_kws={"color": "crimson", "linewidth": 2.5}            
)

plt.title("Synergy Analysis: Math vs Reading Performance", fontsize = 12, fontweight = "bold", pad = 15)
plt.xlabel("Math Score", fontsize = 10, labelpad = 10)
plt.ylabel("Reading Score", fontsize = 10, labelpad = 10)

plt.grid(True, linestyle = "--", alpha = 0.5)
plt.tight_layout()

plt.savefig("../images/math_vs_reading_relationship.png", dpi = 300, bbox_inches = "tight")
plt.show()
```
