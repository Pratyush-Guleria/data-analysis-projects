## 🚻 Gender Distribution

```python
plt.figure(figsize=(5, 5))

sns.countplot(
    data=df,
    x="gender",
    hue="gender",
    palette="Set2",
    legend=False,
    width=0.4,
)


plt.title("Distribution of Students by Gender", fontsize=12, fontweight="bold", pad=15)
plt.xlabel("Gender", fontsize=10, labelpad=10)
plt.ylabel("Number of Students", fontsize=10, labelpad=10) 

plt.tight_layout()
plt.savefig("../images/gender_distribution.png", dpi = 300, bbox_inches = "tight")
plt.show()
```

* **Higher Female Representation:** Female students represent the majority slice of the dataset with **51.8% (518 students)**, slightly outnumbering male students who stand at **48.2% (482 students)**.

* **Data Validity:** With a total sample size of exactly **1,000 observations**, this distribution provides a robust foundation for conducting subsequent cross-feature analysis (such as analyzing scores based on parental education or lunch types).

## 📝 Average Math Score by Parental Level of Education

```python
plt.figure(figsize = (8, 5))

# Sort parental education levels by average math score
education_order = df.groupby("parental level of education")["math score"].mean().sort_values(ascending = False).index

sns.barplot(data = df,
            x = "math score",
            y = "parental level of education",
            order = education_order,
            legend = False,
            errorbar = None,
            hue = "parental level of education",
            palette = "Set2",
            width = 0.5
)
plt.title("Average Math Score by Parental Level of Education", fontsize = 10, fontweight = "bold", pad = 15)

plt.xlabel("Average Math score", fontsize = 10, labelpad = 10)
plt.ylabel("Parental Level of Education", fontsize = 10, labelpad = 10)

plt.tight_layout()

plt.savefig("../images/average_math_score_by_parental_level_of_education.png", dpi = 300, bbox_inches = "tight")
plt.show()
```

### 📊 Key Observations: Average Math Score by Parental Level of Education

* Students whose parents hold bachelor's or master's degrees show the highest average math scores in this dataset.
* Students whose parents have a high school or some high school education exhibit comparatively lower average math scores.
* Overall, the dataset shows a positive association between parental education level and students' average math performance.

##  📊 Correlation Heatmap for Student Academic Scores

```python
plt.figure(figsize=(7, 5.5))

scores_columns = ["math score", "reading score", "writing score"]
correlation_matrix = df[scores_columns].corr()


sns.heatmap(
    correlation_matrix, 
    annot=True,          # Displays the exact correlation numbers inside boxes
    cmap="RdYlBu",       # Distinct color gradient (Red/Yellow/Blue) to show strong vs weak relationships
    fmt=".3f",           # Shows numbers precisely up to 3 decimal places
    vmin=-1, vmax=1,     # Keeps the standard mathematical correlation scale bounds
    linewidths=1,        # Clear white spacing between boxes for crisp visibility
    square=True        # Forces boxes to maintain a perfect square aspect ratio
)

# 💡 PERFECT PROFESSIONAL TITLE
plt.title(
    "Statistical Correlation Analysis: Math, Reading, & Writing Test Scores", 
    fontsize=11, 
    fontweight="bold", 
    pad=20
)

plt.tight_layout()

plt.savefig("../images/correlation_heatmap_student_academic_scores.png", dpi = 300, bbox_inches = "tight")
plt.show()
```
